# Explore the processed dataset
Run `python -m taxlobby build` first. This notebook just loads the outputs so you can explore them.

```python
import pandas as pd
from taxlobby import config
cfg = config.load()
P = cfg['processed_dir']
T = pd.read_csv(P / 'tax_issues.csv.gz', low_memory=False)
M = pd.read_csv(P / 'tax_spend.csv')
people = pd.read_csv(P / 'tax_lobbyists.csv')
firms = pd.read_csv(P / 'tax_firms.csv')
T.shape, M.shape, people.shape, firms.shape
```

```python
# revolver presence by quarter, in-house vs outside
T.assign(rev=T.n_rev > 0).groupby(['quarter', 'side']).rev.mean().unstack()
```

```python
# read some TAX texts by lobbyist type
pd.set_option('display.max_colwidth', 200)
T[['quarter', 'client_name', 'registrant_name', 'side', 'group', 'description']].sample(15, random_state=1)
```
